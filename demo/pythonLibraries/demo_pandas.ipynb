```python
df = pd.read_csv("weather.csv", sep=";")
df.head()
```

```python
import matplotlib.pyplot as plt

df.temp_max.hist()
plt.show()
```

```python
import matplotlib.pyplot as plt

df_plot = df

index = df_plot.date.to_list()
data = df_plot.temp_max.to_list()

plt.plot(index, data)
plt.title("Temperature max en fonction des jours")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2)

axes[0].boxplot(df_plot.temp_min)
axes[0].set_title("Boxplot temperature min")
axes[1].boxplot(df_plot.temp_max)
axes[1].set_title("Boxplot temperature max")
plt.show()
```

```python
df.head()
```

```python
df["year"] = df.date.str.slice(start=0, stop=4)
df["month_number"] = df.date.str.slice(start=5, stop=7)
temperatures = df.groupby(["year", "month_number"]).temp_max.mean().to_list()
index = df.groupby(["year", "month_number"]).temp_max.mean().index.to_list()
index = [b + "-" + a for a, b in index]

fig = plt.figure(figsize=(10, 5))
plt.plot(index, temperatures)
plt.xticks(rotation=60)
plt.title("Evolution de la temperature au cours du temps.")
plt.ylabel("Moyenne T_max mensuelle")
plt.xlabel("mois-annee")

plt.savefig("figure.png")
plt.show()
```
