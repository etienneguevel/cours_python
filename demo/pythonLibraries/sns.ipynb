```python
import pandas as pd

df = pd.read_csv("weather.csv", sep=";")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.lineplot(df, y="temp_max", x="date")
```

```python
sns.histplot(df, x="temp_max")
```

```python
sns.histplot(df, x="temp_max", hue="weather")
```

```python
df["year"] = df["date"].str.slice(0, 4)
df.year
```

```python
sns.violinplot(df, y="temp_max", x="year")
```

```python
sns.pairplot(df, hue="year")
```

```python
sns.relplot(df, x="temp_min", y="temp_max", hue="year", style="weather")
```

```python
fig = plt.figure(figsize=(15, 5))
sns.violinplot(df, y="temp_max", x="year", hue="weather")

plt.show()
```
