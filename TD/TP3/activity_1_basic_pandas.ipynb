## Create a DataFrame object

````python
# Option 1: from a list
data = [("Mary", 968),
        ("Jessica", 155),
        ("Bob", 578),
        ("John", 403),
        ("Mel", 199)]
columns = ["Names", "Births"]

df = pd.DataFrame(data=data, columns=columns)
df
````

````python
# Option 2: from a dictionary
data = {
    "Names": ["Bob", "Jessica", "Mary", "John", "Mel"],
    "Births": [968, 155, 578, 403, 199]
}

df = pd.DataFrame(data=data)
df
````

Dataframes are **tabular data structures**: they contains rows and columns.

The rows and the columns of a dataframe **are both named**. Use the following instructions to get the lists of the row names and the column names:

```python

```

By default, the names of the rows are numbers (0, 1, 2, ...) but it can be modified.

````python
# prints the row and column names
print(df.index)
print(df.columns)
````

## Add columns to a data frame

Adding a new column to an existing data frame is done as follows:

```python

```

if `new_col_name` is the name of the new column and `some_values` is a sequence of the values we want to store in the data frame.

````python
new_col_name = "Number"
values = [10, 11, 12, 13, 14]
df[new_col_name] = values
df
````

````python
df["test_col"] = ["test"]*5
df
````

## Select columns and rows

Selecting both rows and columns is possible knowing the name of the rows and the name of the columns and using the attribute `.loc`:

```python
df.loc[row_names, col_names]
```

where `row_names` is the list of the names of the rows and `col_names` the names of the columns.

````python
data = {
    "Names": ["Bob", "Jessica", "Mary", "John", "Mel"],
    "Births": [968, 155, 578, 403, 199]
}

df = pd.DataFrame(data=data)
````

````python
row_names = [0, 2, 4]
col_names = ["Births"]
df.loc[row_names, col_names]
````

````python
# Or select multiple columns
row_names = [0, 2, 4]
col_names = ["Births", "Names"]
df.loc[row_names, col_names]
````

````python
# Alternativelly
# the character ':' means selecting all the rows of the dataframe
df.loc[:, "Births"]
````

## Sort the data

Sorting data allows you to arrange rows based on the values in one or more columns, making it easier to analyze.

You will learn how to:
1. sort the data frale by a single column
2. sort in descending order
3. sort by multiple columns
4. sort the index

````python
weather[weather.month.isin(["December", "January", "February"])].reset_index(
)
````

````python
weather.sort_values("temp_max", ascending=False).head(10)
````

````python
# Sorting a data frame by a single column

# Create a sample data frame
data = {"Name": ["Alice", "Bob", "Charlie", "David"],
        "Age": [25, 22, 30, 28],
        "Salary": [50000, 60000, 75000, 70000]}
df = pd.DataFrame(data)

# Sort by the "Age" column in ascending order
df_sorted_age = df.sort_values(by="Age")
print(df_sorted_age)
````

````python
# Sorting in descending order

df_sorted_age_desc = df.sort_values(by="Age", ascending=False)
print(df_sorted_age_desc)
````

````python
# Sorting by multiple columns

df_sorted_multiple = df.sort_values(by=["Age", "Salary"], ascending=[True, False])
print(df_sorted_multiple)
````

````python
# Sorting the index, i.e. reorder the rows with respect to the row names

df_sorted_index = df.sort_index()
print(df_sorted_index)
````
