## Working with missing values

Missing values occur when no data value is stored for the variable in an observation.
In general, missing data can occur because of nonresponse, i.e. when no information is provided for one or more items.

It is common in data analysis and can have a significant effect on the conclusions that can be drawn from the data.

So it is important to take care about the missing before performing an analysis.

There are mostly two way to handle missing values:
1. delete the rows or the columns that contains many missing values
2. impute the missing values using the closest observations in the dataset

Usually, missing values are encoded as `NaN` (Not a Number) in Pandas



````python
# Let's have an example

employees = pd.read_csv("data/employees.csv")
employees.head()
````

````python
# A new dataframe with True is the cell is missing
employees.isnull().sum()

# Or True is the cell is NOT missing
# employees.notnull()
````

````python
print("data shape:", employees.shape)

# Remove rows which contain at least one missing value
employees = employees.dropna(axis=0, how="any")
print("data shape:", employees.shape)

# Remove rows which contain only missing values
employees = employees.dropna(axis=0, how="all")
print("data shape:", employees.shape)
````

````python
employees["Team"] = employees["Team"].fillna("Unknown")
employees.head()
````

````python
employees = pd.read_csv("data/employees.csv")


# Missing values imputation
# Is it the best solution?
employees.fillna(0)
````

````python
# Impute the missing values with the last valid observation forward to next valid
# Is it the best solution?
employees.fillna(method="ffill")
````

````python
# Impute on one column and store the result in the data frame
employees["Team"] = employees.Team.fillna("Unknown")
employees
````

**Good practices for missing values imputation**

1. Answer the questions: Do you need to impute the missing values? Do you need to impute the missing values for all the columns?
2. Count the number of missing values per columns. If the number of rows that contains missing values if low, you can juste delete it.
3. Impute the missing values columns by columns because the method used may depend on the type of the column (numerical or categories)

## Working with duplicates

In some cases, the dataset contains exactly identical rows, i.e. duplicates. In a statistical point of view, the results can be biased due to possibly many duplicates.
It is of importance to identify the duplcated rows and to remove these rows.

````python
# Toy example
data = {
    "Names": ["Bob", "Jessica", "Mary", "John", "Mel", "Bob"],
    "Births": [968, 155, 578, 403, 199, 968]
}
df = pd.DataFrame(data=data)
print(df)

# Drop the duplicates and display the resulted dataframe
df.drop_duplicates(["Names"])
````

````python
df.groupby("Names").Births.aggregate(["sum", "mean", "median"])
````

````python
# What if there are pseudo-duplicates, i.e. not exact duplicates
data = {
    "Names": ["Bob", "Jessica", "Mary", "John", "Mel", "bob"],
    "Births": [968, 155, 578, 403, 199, 968]
}
df = pd.DataFrame(data=data)
print(df)

# Try to drop the duplicates and display the resulted dataframe
df.drop_duplicates()
````

**What about duplicated columns?**

Having duplicated columns can rarely occur but it can have a negative effect on your conclusions, for instance if your goal if to identify the most relevant features of your dataset.
So it is important to identify and to understand why they contain the same values. In general, the duplicated columns are simply removed.

## Advanced data selection

Advanced data manipulation involves more sophisticated techniques for selecting, filtering, and transforming data.

````python
import pandas as pd

# Create a sample DataFrame
data = {
    "City": ["New York", "Los Angeles", "Chicago", "Houston"],
    "Population": [8398748, 3980405, 2716000, 2328066],
    "Area": [468.9, 1213.9, 227.3, 637.5],
    "Median_Income": [55350, 60563, 51039, 49882]
}

cities_df = pd.DataFrame(data)
````

````python
# Conditional selection

# Select cities with a population greater than 3 million.
selected_rows = cities_df[cities_df["Population"] > 3000000]
print(selected_rows)
````

````python
# Multiple conditions

# Select cities with a population greater than 2 million and a median income less than $55,000.
selected_conditions = cities_df[
    (cities_df["Population"] > 2000000) 
    & (cities_df["Median_Income"] < 55000)
]
print(selected_conditions)
````

````python
# Conditional selection for categorical columns using String method

# Select cities with names starting with "L".
selected_startswith_L = cities_df[cities_df["City"].str.startswith("L")]
print(selected_startswith_L)
````

````python
# Conditional selection for categorical columns using isin() method

# Example 1: select cities from a predefined list
selected_cities = cities_df[cities_df["City"].isin(["New York", "Los Angeles"])]
print(selected_cities)


# Example 2: select rows based on multiple conditions using the `isin()` method with a data frame.

# First create a list of dictionaries for conditions
conditions = [
    {"City": "New York", "Population": 8398748},
    {"City": "Chicago", "Median_Income": 51039}
]

# Select rows that match any condition
selected_conditions_df = cities_df[cities_df.isin(conditions).all(axis=1)]
print(selected_conditions_df)
````

````python
# Conditional selection using a custom function

# Example : Select cities based on a custom filtering function.
def custom_filter(row):
    return row["Area"] > 300 and row["Median_Income"] > 50000

# Apply the custom function using apply()
# The argument axis specifies the axis along which the function is applied (0 for rows and 1 for columns)
selected_custom_filter = cities_df[cities_df.apply(custom_filter, axis=1)]
print(selected_custom_filter)
````


## Grouping and combining data

Grouping data allows you to split the data into groups based on some criteria, and combining involves aggregating or summarizing the data within each group.

````python
# Grouping data with the groupby() method

data = {"Category": ["A", "B", "A", "B", "A", "B"],
        "Value": [10, 15, 20, 25, 30, 35]}
df = pd.DataFrame(data)

# Group by the "Category" column
grouped = df.groupby("Category")
````

````python
grouped['Value'].sum()
````

````python
# Aggregating data within Groups

# Calculate the sum within each group
sum_per_category = grouped['Value'].sum()
print(sum_per_category)

# Calculate the mean within each group
mean_per_category = grouped['Value'].mean()
print(mean_per_category)
````

````python
# Combining data with agg() method

# e.g. calculate both sum and mean for "Value" within each group
result_agg = grouped.agg({"Value": ["sum", "mean"]})
print(result_agg)
````

````python
# Custom aggregation functions

# Define a custom aggregation function
def custom_agg(values):
    return values.max() - values.min()

# Apply the custom aggregation function within each group
custom_result = grouped.agg({"Value": custom_agg})
print(custom_result)
````

## Concatenating DataFrames

Concatenation is the operation of combining two or more data frames together.
Many concatenation methods exist depending on what you need as a result:
* Horizontal concatenation means stacking the rows of the data frames
* Vertical concatenation means combining the data frames along the columns

````python
# 1/ Vertical concatenation

# Create two sample DataFrames
df1 = pd.DataFrame({"A": [1, 2], "C": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})

# Concatenate vertically (stack rows)
result_vertical = pd.concat([df1, df2], ignore_index=True)
print(result_vertical)

# Remark: the argument ignore_index=True is used to reset the index of the resulting DataFrame.

# 2/ Horizontal concatenation

# Concatenate horizontally (join columns), using the argument axis=1
result_horizontal = pd.concat([df1, df2], axis=1)
print(result_horizontal)
````

````python
# Concatenating dataFrames with different columns

# Create two DataFrames with different columns
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"C": [5, 6], "D": [7, 8]})

# Concatenate vertically with different columns
result_diff_columns = pd.concat([df1, df2], ignore_index=True)
print(result_diff_columns)

# Remark: In this case, Pandas fills in missing values with NaN for columns that don't exist in one of the data frames.
````

````python
# Concatenating data frames with matching columns:

# Create two DataFrames with matching columns
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})

# Concatenate vertically with matching columns
result_match_columns = pd.concat([df1, df2], ignore_index=True)
print(result_match_columns)

# Remark: Here, both DataFrames have the same columns, so the concatenation works seamlessly.
````

## Reshaping and pivoting data

Reshaping and pivoting data involve transforming the structure of a data frame to better suit the analysis or visualization needs.

A pivot table allows to summarize, aggregate, and rearrange tabular data. This is convenient way to transform and organize the data according to their analysis needs.

Components of a pivot table:

- **Rows:** The rows of the pivot table typically represent categorical variables or groups based on which you want to aggregate the data

- **Columns:** The columns represent another set of categorical variables. The data is grouped and summarized based on the intersection of rows and columns

- **Values:** The values in the pivot table are the result of applying an aggregation function (e.g., sum, mean, count) to the numerical data in the original dataset.

A pivot table is also known as cross table and it is related to what we call wide format table.

In other hands, long format tables are the results of the "unpivot" operation: one or more columns are identifier variables (id_vars), while all other columns, considered measured variables (value_vars), are unpivoted to the row axis, leaving just two non-identifier columns, ‘variable’ and ‘value’.

The two main functions for reshaping data are `melt()` (from wide to long format) and `pivot_table()` (from long to wide format).

**Melt Function**

The `melt()` function is used to unpivot or melt a data frame from wide format to long format. It is particularly useful when you have a data frame where columns represent different variables, and you want to transform it to a format where these variables are represented as rows.

````python
import pandas as pd
employees = pd.read_csv("https://gitlabsu.sorbonne-universite.fr/scai/data-visualization/-/raw/main/data/employees.csv").dropna(axis=0)

# melt
melt = pd.melt(employees, id_vars="First Name", var_name="Variable", value_name="Value")
````

````python
# Another example

# rows: observations, cols: features, cells: value for one observation and one feature
data = {
    "Date": ["2022-01-01", "2022-01-02", "2022-01-03"],
    "Temperature": [25, 28, 24],
    "Humidity": [50, 45, 60]
}

df = pd.DataFrame(data)
print(df)

# Long format. cols: ID, Variable, values
melt = pd.melt(df, id_vars=["Date"], var_name="Variable", value_name="Value")
print(melt)

# Wide format from the long format = the initial data
melt.pivot(index="Date", columns="Variable", values="Value")
````

## Working with time series

Time Series Analysis involves working with datasets where the data points are indexed by time, see the example below. Pandas provides a set of tools for handling and analyzing time series data:

* Convert the dates to a convinient format
* Resample the data: upsampling or downsampling.
* Rolling statistics
* Time-based slicing: selecting a range of data for a time interval

```yaml
      Date  Temperature  Humidity
2022-01-01           25        50
2022-01-02           28        45
2022-01-03           24        60
...                 ...       ...
```

````python
# Toy dataset

import pandas as pd

data = {
    "date": ["2022-01-01", "2022-01-02", "2022-01-03"],
    "temperature": [25, 28, 24],
    "humidity": [50, 45, 60]
}

df = pd.DataFrame(data)
df
````

````python
df.dtypes
````

**Converting to data frame and setting index**

This is to ensure that the time-related columns are in datetime format and set as the index. This allows for convenient time-based operations.


````python
# Convert "Date" to datetime format
df["date"] = pd.to_datetime(df["date"])
df.dtypes
````

````python
df.dtypes
````

````python
df
````

````python
df["date"].dt.strftime("%m/%Y")
````

````python
# The datetime object implements usefull methods, e.g. that extract the year, month, day, etc.
dt = df["date"]
year = dt.dt.year
month = dt.dt.month_name()
day = dt.dt.day_name()

# Print the results
print(year)
print(month)
print(day)
````

````python
# set the datetime object as the index
df.set_index("date", inplace=True) # inplace modify the data frame and does not return anything
df
````

**Resampling**

This involves changing the frequency of the time series data.
The function `resample()` is used to aggregate or downsample the data based on a specified frequency (e.g., daily to monthly).

````python
# Resample daily data to monthly and calculate the mean
df_resampled = df.resample("M").mean()

df_resampled
````

**Rolling Statistics:**

Rolling functions help in calculating rolling statistics like moving averages or rolling standard deviations.

````python
# Calculate a 3-day rolling mean
df["Rolling_Mean"] = df["temperature"].rolling(window=3).mean()
df
````

**Time-based Slicing**

Pandas allows you to slice and filter data based on time intervals.

````python
# Select data for a specific time range
df_subset = df["2022-01-01":"2022-01-02"]
df_subset
````
