## Vector

Implémenter une class Vector qui stocke l'information d'un vecteur de `n` valeurs initialement stockée dans une liste. Exemple :

```python
>>> vector = Vector([1, 2, 3])
>>> vector.size
3
>>> vector.norm()
3.74
```

Modifier la classe pour gérer plusieurs structures de données (`dict`, `tuples`, etc.), exemple :

```python
>>> vector = Vector([1, 2, 3])
>>> vector = Vector({"a": 1, "b": 2, "c": 3})
>>> vector = Vector((1, 2, 3))
```

````python
# Implementez ici
class Vector:
    def __init__(self, data):
        self.data = data

    @property
    def size(self):
        return len(self.data)

    def norm(self):
        return round(sum([i**2 for i in self.data])**0.5, 2)
    
vector = Vector([1, 2, 3])
print(vector.size)
print(vector.norm())

vector.data.append(4)
print(vector.data)

print(vector.size)
print(vector.norm())
````
