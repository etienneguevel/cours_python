## Définition d'une classe

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
```

```python
point = Point(1, 2)
print(point)
print(type(point))
print(id(point))
```

```python
point2 = Point(1, 2)
print(point2)
print(type(point2))
print(id(point2))
print(point is point2)
```

```python
p1 = Point(1, 2)
p2 = Point(2, 1)
print(p1.point_type)
print(p2.point_type)
```

## Méthode d'instance

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def norm(self):
        return math.sqrt(self.x**2 + self.y**2)
```

```python
p1 = Point(1, 2)
p2 = Point(2, 1)
print(p1.norm())
print(p2.norm())
```

## Méthode de classe

```python
class Point:
    type = "2D"
    def __init__(self, x, y):
        self.x = x
        self.y = y

    @classmethod
    def print_type(cls):
        print(f"My type is {cls.type}")
```

```python
Point.print_type()
```

```python
p1 = Point(1, 2)
p1.print_type()
```

## Méthode statique

```python
import math

class Point:
    type = "2D"
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def norm(self):
        return self.norm2d(self.x, self.y)

    @staticmethod
    def norm2d(x, y):
        return math.sqrt(x**2 + y**2)
```

```python
p1 = Point(1, 2)
print(p1.norm())
print(p1.norm2d(1, 2))
```

### Décorateur `property`

Le décorateur `property` permet l'appel de méthodes d'instance de la même manière qu'un attribut. C'est une autre façon de définir des accesseurs (ou getters).


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    @property
    def x(self):
        return self._x
    
    @x.setter
    def x(self, value):
        self._x = value
```

```python
point = Point(1, 2)
print(point.x)
print(point._x)

point.x = 10
print(point.x)
```

### Rendre les attributs "cachés"

On peut protéger dans une certaine mesure les attributs avec `@property`
en ne définissant pas de setter.

```python
class Point():
    def __init__(self, x, y):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        return self.__x
```

```python
point = Point(1, 2)
print(point.x)
```

```python
# Essayons de modifier la valeur de x
point.x = 2
```

```python
# Rappelons que l'attribut __x n'est pas accessible, encore moins modifiable
point.__x
```

```python
# Mais on peut voir __x avec la fonction dir

dir(point)
```

```python
# Et modifier l'attribut avec le nom correct

point._Point__x = 2
point.x
```
