### Etudiants

```python
class Person:
    def __init__(self, firstname, lastname):
        self.__firstname = firstname
        self.__lastname = lastname

    def __str__(self):
        return f"{self.firstname} {self.lastname}"
    
    @property
    def firstname(self):
        return self.__firstname
    
    @property
    def lastname(self):
        return self.__lastname
```

```python
class Student(Person):
    def __init__(self, firstname, lastname, university, student_id):
        super().__init__(firstname, lastname)
        self.__university = university
        self.id = student_id

    @property
    def university(self):
        return self.__university
    
    def __str__(self):
        return (f"{self.firstname} {self.lastname}")+ " | " + (f"Student ID: {self.id}")
```

```python
eleve = Student("Etienne", "Guevel", "Sorbonne", 123456)
print(eleve)
```

Implémenter une classe `Teacher` sur le même modèle que `Student` avec des
attributs spécifiques (`teacher_id`, `salary`, `courses`) un `__str__` modifié.


```python
class Teacher(Person):
    def __init__(self, firstname, lastname, university, salary, teacher_id, courses):
        super().__init__(firstname, lastname)
        self.__university = university
        self.salary = salary
        self.teacher_id = teacher_id
        self.courses = courses
    
    @property
    def university(self):
        return self.__university
    
    def __str__(self):
        return (f"{self.firstname} {self.lastname}")+ " | " + (f"Teacher ID: {self.teacher_id}")
```

```python
class Teacher(Student):
    def __init__(self, firstname, lastname, university, salary, teacher_id, courses):
        super().__init__(firstname, lastname, university, teacher_id)
        self.salary = salary
        self.courses = courses
    
    def __str__(self):
        return (f"{self.firstname} {self.lastname}")+ " | " + (f"Teacher ID: {self.id}")
```

```python
teacher = Teacher("Etienne", "Guevel", "Sorbonne", 40000, 123456, "python")
print(teacher)
```

### Polygones

1. Implémenter une classe `Polygon` qui contient le nombre de côtés d'un polygone `nr_sides` et la liste de la longueur de chaque côté `sides`.  
Du style : 
`polygon = Polygon(sides=[1, 1, 1, 1, 1])`  
2. Implémenter une classe `Rectangle` qui hérite de la classe `Polygon` et qui contient une méthode `area` permettant de calculer l'aire.
3. Implémenter une classe `Square`

```python
class Polygon:
    def __init__(self, sides):
        self.sides = sides

    @property
    def nr_sides(self):
        return len(self.sides)
    
    def __repr__(self):
        return f"Polygon avec {self.nr_sides} cotes : {self.sides}"
    
```

```python
polygon = Polygon(sides=[1, 1, 1, 1])
polygon
```

```python
class Rectangle(Polygon):
    def __init__(self, sides):
        if len(sides) == 2:
            longueur, largeur = sides
            super().__init__(sides=[longueur, largeur, longueur, largeur])
            self.longueur = longueur
            self.largeur = largeur
        
        elif len(sides) == 4:
            n_sides = len(set(sides))
            if n_sides > 2:
                raise ValueError(f"{n_sides} uniques ont ete donne, 1 ou 2 sont les seules valeurs possibles.")
            
            super().__init__(sides=sides)
            if n_sides == 1:
                self.longueur = sides[0]
                self.largeur = sides[0]
            
            else:
                self.longueur, self.largeur = set(sides)

        else:
            raise ValueError(f"{len(sides)} cotes ont ete donnes, 2 ou 4 sont les valeurs acceptes.")

    @property
    def area(self):
        return self.largeur * self.longueur
```

```python
rectangle = Rectangle([2, 2, 2, 2])
print(rectangle)
```

```python
class Square(Rectangle):
    def __init__(self, sides):
        if len(sides) == 1:
            longueur = sides[0]
            super().__init__(sides=[longueur, longueur])
        
        elif len(sides) == 4:
            n_sides = len(set(sides))
            if n_sides != 1:
                raise ValueError(f"{n_sides} uniques ont ete donne, 1 ou 2 sont les seules valeurs possibles.")
            
            super().__init__(sides=sides)
```
